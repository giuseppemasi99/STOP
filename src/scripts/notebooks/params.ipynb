```python
import json
from itertools import product
import os
```

```python
def get_path(model, method, n_m, max_d, id, C_avg=None, R_avg=None):
    path = f'results_retry/{model}/{method}/n_m={n_m}-max_d={max_d}'
    if 'STOP' in method or 'Li' in method:
        path += f'/C_avg={C_avg}-R_avg={R_avg}'
    path += f'/id={id}'
    return path
```

```python
def exist(path):
    file = f'{path}/t=299.json'
    return os.path.exists(file)
```

```python
seeds = range(30)
model = 'Yolo'
baselines = ['Random', 'RoundRobin']
methods = ['STOP-V=1000']
n_m = range(3, 8)
max_d = [1000]
C_avg = [.003]
R_avg = [7]
l = list()
for s, m, n, d in product(seeds, baselines, n_m, max_d):
    if not exist(get_path(model, m, n, d, s)):
        l.append((m, model, n, d, s, -1, -1, -1))
for s, m, n, d, c, r in product(seeds, methods, n_m, max_d, C_avg, R_avg):
    if not exist(get_path(model, m, n, d, s, c, r)):
        m, V = m.split('-V=')
        l.append((m, model, n, d, s, c, r, float(V)))
print(len(l))
with open("params_retry.txt", "w") as f:
    for item in l:
        f.write(' '.join(str(x) for x in item) + '\n')
```
