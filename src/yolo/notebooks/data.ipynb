```python
import os
import matplotlib.pyplot as plt
```

```python
data_path = 'animals/complete'
```

```python
animal2count_train = dict()
for animal in os.listdir(f'{data_path}/train'):
    animal2count_train[animal] = len(os.listdir(f'{data_path}/train/{animal}'))
animal2count_train = dict(sorted(animal2count_train.items(), key=lambda item: item[1]))
```

```python
animal2count_test = dict()
for animal in os.listdir(f'{data_path}/val'):
    animal2count_test[animal] = len(os.listdir(f'{data_path}/val/{animal}'))
animal2count_test = dict(sorted(animal2count_test.items(), key=lambda item: item[1]))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 15))

axes[0].barh(list(animal2count_train.keys()), animal2count_train.values())
axes[0].set_title('Train')

axes[1].barh(list(animal2count_test.keys()), animal2count_test.values())
axes[1].set_title('Val')
plt.show()
plt.close()
```

```python
import cv2 as cv
import numpy as np
```

```python
train_o_val = 'train'
npixels2fsize = dict()

for animal in os.listdir(f'{data_path}/{train_o_val}'):
    for fname in os.listdir(f'{data_path}/{train_o_val}/{animal}'):
        img = cv.imread(f'{data_path}/{train_o_val}/{animal}/{fname}')
        file_stats = os.stat(f'{data_path}/{train_o_val}/{animal}/{fname}')
        if 'jpg' in fname:
            npixels2fsize[img.shape[0]*img.shape[1]] = file_stats.st_size
```

```python
print('Min pixels = ', min(npixels2fsize))
print('Max pixels = ', max(npixels2fsize))
```

```python
ratios = [p/s for p, s in npixels2fsize.items()]
min(ratios), max(ratios)
```
